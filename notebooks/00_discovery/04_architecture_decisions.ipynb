# Enterprise Architecture Decisions

## Purpose

This document defines the main architecture decisions derived from the initial data discovery phase.

The objective is to establish the foundation for the future Enterprise Data Platform implementation, including:

- Lakehouse architecture approach
- Data modeling strategy
- Data product definition
- Governance principles

The decisions documented here are based on the following technical artifacts:

- Enterprise Data Inventory
- Data Profiling Report
- Enterprise Relationship Discovery

These artifacts provide the technical evidence used to support architecture decisions.

---

# 1. Architecture Context

## Current State Assessment

The initial discovery phase identified multiple business domains:

- Commerce
- Customer
- Product
- Seller
- Delivery

The source landscape contains:

- Transactional datasets
- Master data entities
- Reference data entities

The discovered relationships provide the foundation for defining enterprise business entities, analytical models, and future data products.

---

# 2. Lakehouse Architecture Decision

## Decision

Adopt a Medallion Lakehouse Architecture as the target data platform design.

The platform will organize data into three logical layers:

Bronze
|
Silver
|
Gold


## Rationale

The source datasets require different levels of processing and consumption.

The layered approach provides:

- Source preservation
- Data quality management
- Business-ready analytical models
- Improved governance and traceability

---

# 3. Bronze Layer Decision

## Purpose

Store source data with minimal transformation while preserving the original structure.

## Characteristics

- Preserve source structure
- Maintain ingestion history
- Enable data replay
- Support auditing

## Initial Domains
Bronze
  - commerce/raw
  - customer/raw
  - delivery/raw


---

# 4. Silver Layer Decision

## Purpose

Create standardized and trusted business entities.

## Transformations

The Silver layer will apply:

- Schema standardization
- Data type corrections
- Data quality validation
- Duplicate management
- Business rule application

## Expected Entities
Silver
  - Customers
  - Orders
  - Order Items
  - Products
  - Sellers
  - Geolocation


---

# 5. Gold Layer Decision

# 8. Semantic Relationship Decision

## Decision

Maintain semantic relationships where different physical columns represent the same business concept.

### Identified Example

Postal code relationship:
customer_zip_code_prefix
    |

geolocation_zip_code_prefix

and:

seller_zip_code_prefix
    |
geolocation_zip_code_prefix

## Rationale

These relationships enable geographic enrichment and support location-based analytical models.
